```python
import sys
sys.path.append('..')
from epicast.utils import *
pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)
```

```python
metadata = pd.read_csv('../data/AlphaGenome/AlphaGenome_metadata.csv')
metadata
```

```python
metadata['Assay title'].value_counts()
```

```python
metadata['cell_type'] = metadata['biosample_name']
metadata['assay'] = metadata['Assay title']

mask = (metadata['Assay title'] == 'DNase-seq')
metadata.loc[mask, 'assay'] = 'DNase'
mask = (metadata['Assay title'] == 'TF ChIP-seq')
metadata.loc[mask, 'assay'] = metadata.loc[mask, 'transcription_factor']
mask = (metadata['Assay title'] == 'Histone ChIP-seq')
metadata.loc[mask, 'assay'] = metadata.loc[mask, 'histone_mark']

metadata['index'] = metadata.index
```

```python
df_pivot = metadata.pivot_table(
    values="index", 
    index="cell_type", 
    columns="assay", 
    aggfunc=list,
)

df_pivot
```

```python
assays = ['DNase', 'H3K4me3', 'H3K27ac', 'CTCF']
df_pivot.loc[df_pivot[assays].notna().all(axis=1), assays]
```

```python
cell_types = ['HEK293T', 'HeLa-S3', 'HepG2', 'HL-60', 'Jurkat', 'K562', 'MCF-7', 'PANC-1', 'Raji']
cell_names = ['HEK293T', 'HeLa-S3', 'HepG2', 'HL-60', 'Jurkat', 'K562', 'MCF-7', 'Panc1', 'Raji', 'DND-41', 'CD4-positive, alpha-beta T cell', 'CD8-positive, alpha-beta T cell', 'GM12878','B cell']

assays = ['DNase', 'H3K4me3', 'H3K27ac', 'CTCF']

df_track = df_pivot.loc[cell_names, assays]
df_track
```

```python
pred_DNase = load_h5('../predict_CRE_activity/outputs/TFBU_MPRA_18000_AlphaGenome_pred_DNase.h5')
print(pred_DNase.shape)

pred_Histone = load_h5('../predict_CRE_activity/outputs/TFBU_MPRA_18000_AlphaGenome_pred_Histone.h5')
print(pred_Histone.shape)

pred_TF = load_h5('../predict_CRE_activity/outputs/TFBU_MPRA_18000_AlphaGenome_pred_TF.h5')
print(pred_TF.shape)
```

```python
MPRA_df = pd.read_csv('../data/TFBU_9cell/data.tsv', sep='\t')
MPRA_df.describe()
```

```python
notna_indice = MPRA_df[MPRA_df[cell_types].notna().all(axis=1)].index
MPRA_df = MPRA_df.loc[notna_indice]
```

```python
data = MPRA_df.drop(columns=['id', 'seq'])
data = (data - data.mean(axis=0)) / data.std(axis=0)
data.describe()
```

```python
mean = data.mean(axis=1)  # 每条序列跨细胞均值（忽略 NaN）
for cell_type in cell_types:
    diff = (data[cell_type] - mean).abs()
    thr = diff.quantile(0.8)
    mask = (diff >= thr)
    data[f"{cell_type}_specific"] = mask
```

```python
VEF_df = pd.DataFrame()

for i, cell_name in enumerate(cell_names):
    for j, assay in enumerate(assays):
        indice = df_track.loc[cell_name, assay]
        if isinstance(indice, list) and len(indice) > 0:
            if assay == 'DNase':
                indice = [i-713 for i in indice]
                pred_array = pred_DNase
            elif assay in ['H3K4me3', 'H3K27ac']:
                indice = [i-1685 for i in indice]
                pred_array = pred_Histone
            elif assay == 'CTCF':
                indice = [i-2803 for i in indice]
                pred_array = pred_TF
            else:
                VEF_df[f'{cell_name}_{assay}'] = np.nan
                continue
            pred = pred_array[:, indice].mean(1)
            VEF_df[f'{cell_name}_{assay}'] = pred
        else:
            VEF_df[f'{cell_name}_{assay}'] = np.nan

VEF_df.describe()
```

```python
VEF_df = VEF_df.loc[notna_indice]
```

```python
DNase_cols = [f'{cell_name}_DNase' for cell_name in cell_names]
VEF_df[DNase_cols] = VEF_df[DNase_cols] * 10
other_cols = [col for col in VEF_df.columns if col not in DNase_cols]
VEF_df[other_cols] = VEF_df[other_cols] / 10

VEF_df = np.log1p(VEF_df)
VEF_df.describe()
```

```python
for j, assay in enumerate(assays):
    pearson_df = pd.DataFrame()
    for i, cell_type in enumerate(cell_types):
        for ii, cell_name in enumerate(cell_names):
            pred = VEF_df[f'{cell_name}_{assay}']
            true = MPRA_df[f'{cell_type}']
            r, _ = pearson(pred, true)
            pearson_df.loc[f'{cell_name}_{assay}', f'{cell_type}'] = r
    print(pearson_df)
```

```python
for j, assay in enumerate(assays):
    pearson_df = pd.DataFrame()
    for i, cell_type in enumerate(cell_types):
        for ii, cell_name in enumerate(cell_names):
            pred = VEF_df[f'{cell_name}_{assay}']
            true = data[f'{cell_type}']

            indice = data[f'{cell_type}_specific']
            pred = pred.loc[indice]
            true = true.loc[indice]
            r, _ = pearson(pred, true)
            pearson_df.loc[f'{cell_name}_{assay}', f'{cell_type}'] = r
    print(pearson_df)
```

```python
# 每个 assay 固定 scale factor

VEF_df = pd.read_csv('../data/TFBU_MPRA/TFBU_MPRA_AG_VEF_raw.tsv', sep='\t')

cell_types = ['HepG2', 'HEK293T', 'Jurkat', 'PANC']
assays = ['DNase', 'H3K4me3', 'H3K27ac', 'CTCF']

for assay in assays:
    for cell_type in cell_types:
        if assay == 'DNase':
            scale_factor = 0.1
        else:
            scale_factor = 10
        x = VEF_df[f'{cell_type}_{assay}'].copy()
        x = np.log1p(x / scale_factor)
        VEF_df[f'{cell_type}_{assay}'] = x

VEF_df.to_csv('../data/TFBU_MPRA/TFBU_MPRA_AG_VEF_log1p_scale.tsv', sep='\t', index=False)

pearson_df = pd.DataFrame(dtype=float)
for i, cell_type in enumerate(cell_types):
    for j, assay in enumerate(assays):
        pred = VEF_df[f'{cell_type}_{assay}']
        true = MPRA_df[f'{cell_type}']
        r, _ = pearson(pred, true)
        pearson_df.loc[cell_type, assay] = r
print(pearson_df)
```
